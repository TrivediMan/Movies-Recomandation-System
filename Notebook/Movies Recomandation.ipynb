```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
base_path = Path().cwd().parent
model_path = base_path/"Model"
movie_path = base_path/"Data"/"tmdb_5000_movies.csv"
credit_path = base_path/"Data"/"tmdb_5000_credits.csv"
```

```python
DF_mov = pd.read_csv(movie_path)
DF_mov
```

```python
DF_cre = pd.read_csv(credit_path)
DF_cre
```

```python
DF_mov = DF_mov.merge(DF_cre,on='title')
```

```python
DF_mov.head()
```

```python
DF_mov = DF_mov[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
DF_mov.head()
```

```python
import ast
```

```python
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name']) 
    return L 
```

```python
DF_mov.dropna(inplace=True)
```

```python
DF_mov['genres'] = DF_mov['genres'].apply(convert)
DF_mov.head()
```

```python
DF_mov['keywords'] = DF_mov['keywords'].apply(convert)
DF_mov.head()
```

```python
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
        counter+=1
    return L 
```

```python
DF_mov['cast'] = DF_mov['cast'].apply(convert)
DF_mov.head()
```

```python
DF_mov['cast'] = DF_mov['cast'].apply(lambda x:x[0:3])
```

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L
```

```python
DF_mov['crew'] = DF_mov['crew'].apply(fetch_director)
```

```python
DF_mov.sample(5)
```

```python
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
DF_mov['cast'] = DF_mov['cast'].apply(collapse)
DF_mov['crew'] = DF_mov['crew'].apply(collapse)
DF_mov['genres'] = DF_mov['genres'].apply(collapse)
DF_mov['keywords'] = DF_mov['keywords'].apply(collapse)
```

```python
DF_mov.head()
```

```python
DF_mov['overview'] = DF_mov['overview'].apply(lambda x:x.split())
```

```python
DF_mov['tags'] = DF_mov['overview'] + DF_mov['genres'] + DF_mov['keywords'] + DF_mov['cast'] + DF_mov['crew']
```

```python

new = DF_mov.drop(columns=['overview','genres','keywords','cast','crew'])
```

```python
new['tags'] = new['tags'].apply(lambda x: " ".join(x))
new.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
    
```

```python
vector = cv.fit_transform(new['tags']).toarray()
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
new[new['title'] == 'The Lego Movie'].index[0]
```

```python
def recommend(movie):
    index = new[new['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(new.iloc[i[0]].title)
```

```python
recommend('Gandhi')
```
